```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df_train = pd.read_csv("train.csv", index_col=0)
df_test = pd.read_csv("test.csv", index_col=0)
df_train
```

```python
features = list(set(df_train.columns) - {"zipcode", "price"})
```

```python
# preprocessing
df_train["price"] /= 1000
df_test["price"] /= 1000
mean = df_train.mean()
std = df_train.std()
df_train = (df_train - mean) / std
df_test = (df_test - mean) / std
```

```python
X_train = df_train[features]
y_train = df_train["price"]

model_train = LinearRegression().fit(X_train, y_train)
print("Coefficients:\n", *[f"{f.ljust(15)}: {c}\n" for f, c in zip(features, model_train.coef_)])
print("Intercept: ", model_train.intercept_)

y_train_pred = model_train.predict(X_train)

train_mse = mean_squared_error(y_train, y_train_pred)
train_r2 = r2_score(y_train, y_train_pred)

print("Training MSE:", train_mse)
print("Training R2:", train_r2)
```

```python
X_test = df_test[features]
y_test = df_test["price"]

model_test = LinearRegression().fit(X_test, y_test)
print("Coefficients:\n", *[f"{f.ljust(15)}: {c}\n" for f, c in zip(features, model_test.coef_)])
print("Intercept: ", model_test.intercept_)

y_test_pred = model_test.predict(X_test)

test_mse = mean_squared_error(y_test, y_test_pred)
test_r2 = r2_score(y_test, y_test_pred)

print("Testing MSE:", test_mse)
print("Testing R2:", test_r2)
```

```python
def poly_reg(x, p):
    X = np.ones((len(x), p + 1))
    for i in range(1, p + 1):
        X[:, i] = X[:, i - 1] * x
    return X

for p in range(1, 5):
    x, y = poly_reg(X_train["sqft_living"], p), y_train
    b = np.matmul(np.matmul(np.linalg.inv(np.matmul(x.T, x)), x.T), y)
    predictions = np.matmul(x, b)

    mse = np.mean((y - predictions) ** 2)
    print(f"Polynomial Regression Train MSE (p = {p}):", mse)

    R2 = 1 - np.sum((y - predictions) ** 2) / np.sum((y - np.mean(y)) ** 2)
    print(f"Polynomial Regression Train R2 (p = {p}):", R2)

    x, y = poly_reg(X_test["sqft_living"], p), y_test
    predictions = np.matmul(x, b)
    mse = np.mean((y - predictions) ** 2)
    print(f"Polynomial Regression Test MSE (p = {p}):", mse)

    R2 = 1 - np.sum((y - predictions) ** 2) / np.sum((y - np.mean(y)) ** 2)
    print(f"Polynomial Regression Test R2 (p = {p}):", R2)

    print()
    
```

```python
def gradient_descent(X, y, lr, iterations):
    n, m = tuple(X.shape)
    theta = np.zeros(m)

    for _ in range(iterations):
        predictions = np.matmul(X, theta)
        grad = (2 / n) * np.matmul(X.T, (predictions - y))
        theta -= lr * grad

    return theta

for lr in (0.01, 0.1, 0.5):
    for iterations in (10, 50, 100):
        X = np.vstack([np.ones(len(X_train)), X_train.T]).T
        y = np.array(y_train)

        theta = gradient_descent(X, y, lr, iterations)

        train_pred = np.matmul(X, theta)
        train_mse = mean_squared_error(y, train_pred)
        train_r2 = r2_score(y, train_pred)

        X = np.vstack([np.ones(len(X_test)), X_test.T]).T
        y = np.array(y_test)

        test_pred = np.matmul(X, theta)
        test_mse = mean_squared_error(y, test_pred)
        test_r2 = r2_score(y, test_pred)

        print(f"Gradient Descent (lr = {lr}, iterations = {iterations}) Train MSE: {train_mse}")
        print(f"Gradient Descent (lr = {lr}, iterations = {iterations}) Train R2: {train_r2}")
        print(f"Gradient Descent (lr = {lr}, iterations = {iterations}) Test MSE: {test_mse}")
        print(f"Gradient Descent (lr = {lr}, iterations = {iterations}) Test R2: {test_r2}")
```

```python
def ridge_gradient_descent(X, y, lr, iterations, lmbda):
    n, m = tuple(X.shape)
    theta = np.zeros(m)

    for _ in range(iterations):
        predictions = np.matmul(X, theta)
        grad = (2 / n) * np.matmul(X.T, (predictions - y)) + 2 * lmbda * theta
        theta -= lr * grad

    return theta

lr = 0.001
iterations = 100
N = 1000
X = np.random.uniform(-2, 2, N)
e = np.random.normal(0, 2, N)
y = 1 + 2 * X + e
X = np.vstack([np.ones(N), X]).T

for lmbda in (1, 10, 100, 1000, 10000):
    theta = ridge_gradient_descent(X, y, lr, iterations, lmbda)

    train_pred = np.matmul(X, theta)
    train_mse = mean_squared_error(y, train_pred)
    train_r2 = r2_score(y, train_pred)

    print(f"Ridge Gradient Descent (lr = {lr}, iterations = {iterations}, lambda = {lmbda}) Train MSE: {train_mse}")
    print(f"Ridge Gradient Descent (lr = {lr}, iterations = {iterations}, lambda = {lmbda}) Train R2: {train_r2}")
```
