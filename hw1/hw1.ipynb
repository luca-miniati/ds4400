```python
import pandas as pd
import numpy as np
```

```python
fn = "kc_house_data.csv"
df = pd.read_csv(fn)
df["neg_price"] = -df["price"]
```

```python
features = list(set(df.columns) - {"id", "date", "zipcode"})
features
```

```python
stats = pd.DataFrame(
    [
        {
            "feature": f,
            "mean": df[f].mean(),
            "min": min(df[f]),
            "max": max(df[f]),
            "variance": np.var(df[f]),
        }
        for f in features
    ]
)
stats
```

```python
print(stats.loc[stats["mean"].idxmin()])
print()
print(stats.loc[stats["mean"].idxmax()])
```

```python
print(stats.loc[stats["variance"].idxmin()])
print()
print(stats.loc[stats["variance"].idxmax()])
```

```python
corr = df[features].corr()["price"].sort_values()
corr
```
